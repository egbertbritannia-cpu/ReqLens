```python
# --- ELECTRA WITH Augmentation: split first, augment train only ---
import os
os.environ["WANDB_DISABLED"] = "true"
os.environ["TRANSFORMERS_NO_TF"] = "1"
os.environ["TRANSFORMERS_NO_FLAX"] = "1"

import copy
import random
import re

import nltk
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import seaborn as sns

from nltk.corpus import stopwords, wordnet
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support,
)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from torch.utils.data import DataLoader, Dataset
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
    get_linear_schedule_with_warmup,
)

# --- Config ---
RANDOM_STATE = 42
MODEL_NAME = "google/electra-base-discriminator"
DATA_URL = "https://raw.githubusercontent.com/egbertbritannia-cpu/rbl-requirements-classification/main/data/exp/promise_exp.csv"

MAX_LENGTH = 128
NUM_EPOCHS = 30
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
TRAIN_BATCH_SIZE = 16
EVAL_BATCH_SIZE = 32
WARMUP_RATIO = 0.1

AUGMENTATION_TARGETS = {
    "FT": 200,
    "L": 200,
    "MN": 200,
    "A": 200,
    "LF": 200,
    "SC": 200,
    "PO": 200,
    "PE": 200,
    "O": 200,
    "US": 200,
    "SE": 100,
}

KEEP_WORDS = {"not", "no", "shall", "must", "should", "will", "cannot", "never", "only"}

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
stop_words = set(stopwords.words("english")) - KEEP_WORDS


def get_synonyms(word):
    clean_word = re.sub(r"[^a-zA-Z]", "", str(word).lower())
    if not clean_word:
        return []
    synonyms = set()
    for syn in wordnet.synsets(clean_word):
        for lemma in syn.lemmas():
            synonym = lemma.name().replace("_", " ").lower()
            if synonym != clean_word:
                synonyms.add(synonym)
    return list(synonyms)


def synonym_replacement(text, n=1):
    words = str(text).split()
    if not words:
        return str(text)

    new_words = words.copy()
    candidates = list({
        word for word in words
        if re.sub(r"[^a-zA-Z]", "", word).lower() not in stop_words and len(word) > 3
    })
    random.shuffle(candidates)

    replaced = 0
    for word in candidates:
        synonyms = get_synonyms(word)
        if synonyms:
            synonym = random.choice(synonyms)
            new_words = [synonym if current == word else current for current in new_words]
            replaced += 1
        if replaced >= n:
            break

    return " ".join(new_words)


def phrase_insertion(text, n=1):
    phrases = ["the system shall", "must be able to", "should ensure that"]
    words = str(text).split()
    if not words:
        return str(text)

    for _ in range(n):
        insert_at = random.randint(0, len(words) - 1)
        words.insert(insert_at, random.choice(phrases))

    return " ".join(words)


def random_swap(text, n=1):
    words = str(text).split()
    if len(words) < 2:
        return str(text)

    for _ in range(n):
        i, j = random.sample(range(len(words)), 2)
        words[i], words[j] = words[j], words[i]

    return " ".join(words)


def random_deletion(text, p=0.1):
    words = str(text).split()
    if len(words) <= 1:
        return str(text)

    remaining = [word for word in words if random.random() > p]
    return " ".join(remaining) if remaining else random.choice(words)


def augment_requirement_text(text):
    text = synonym_replacement(text, n=1)
    text = phrase_insertion(text, n=1)
    text = random_swap(text, n=1)
    text = random_deletion(text, p=0.1)
    return " ".join(str(text).split())


class RequirementsDataset(Dataset):
    def __init__(self, frame, tokenizer):
        self.labels = frame["label"].astype(int).tolist()
        self.encodings = tokenizer(
            frame["model_text"].astype(str).tolist(),
            truncation=True,
            padding="max_length",
            max_length=MAX_LENGTH,
        )

    def __getitem__(self, idx):
        item = {key: torch.tensor(values[idx]) for key, values in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

    def __len__(self):
        return len(self.labels)


def compute_metrics_from_arrays(labels, preds):
    weighted_p, weighted_r, weighted_f1, _ = precision_recall_fscore_support(
        labels, preds, average="weighted", zero_division=0
    )
    macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
        labels, preds, average="macro", zero_division=0
    )
    return {
        "accuracy": accuracy_score(labels, preds),
        "precision": weighted_p,
        "recall": weighted_r,
        "f1": weighted_f1,
        "precision_macro": macro_p,
        "recall_macro": macro_r,
        "macro_F1": macro_f1,
    }


@torch.no_grad()
def evaluate_model(model, data_loader, device):
    model.eval()
    all_labels = []
    all_preds = []
    total_loss = 0.0

    for batch in data_loader:
        batch = {key: value.to(device) for key, value in batch.items()}
        outputs = model(**batch)
        total_loss += outputs.loss.item()

        preds = outputs.logits.argmax(dim=-1)
        all_labels.extend(batch["labels"].detach().cpu().numpy())
        all_preds.extend(preds.detach().cpu().numpy())

    metrics = compute_metrics_from_arrays(np.array(all_labels), np.array(all_preds))
    metrics["loss"] = total_loss / max(1, len(data_loader))
    return metrics, np.array(all_labels), np.array(all_preds)


def compute_class_weights(frame, label_encoder):
    classes = np.arange(len(label_encoder.classes_))
    weights = compute_class_weight(
        class_weight="balanced",
        classes=classes,
        y=frame["label"].astype(int).to_numpy(),
    )
    return torch.tensor(weights, dtype=torch.float)


def train_electra(model, train_loader, val_loader, device, class_weights):
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
    )

    total_steps = max(1, NUM_EPOCHS * len(train_loader))
    warmup_steps = int(WARMUP_RATIO * total_steps)

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_steps,
    )

    loss_fct = torch.nn.CrossEntropyLoss(weight=class_weights.to(device))
    best_macro_F1 = -1.0
    best_state_dict = None

    for epoch in range(1, NUM_EPOCHS + 1):
        model.train()
        total_train_loss = 0.0

        for batch in train_loader:
            batch = {key: value.to(device) for key, value in batch.items()}
            labels = batch.pop("labels")

            optimizer.zero_grad(set_to_none=True)
            outputs = model(**batch)
            loss = loss_fct(outputs.logits, labels)

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            scheduler.step()

            total_train_loss += loss.item()

        val_metrics, _, _ = evaluate_model(model, val_loader, device)
        avg_train_loss = total_train_loss / max(1, len(train_loader))

        if val_metrics["macro_F1"] > best_macro_F1:
            best_macro_F1 = val_metrics["macro_F1"]
            best_state_dict = copy.deepcopy({
                key: value.detach().cpu()
                for key, value in model.state_dict().items()
            })

        print(
            f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
            f"train_loss={avg_train_loss:.4f} | "
            f"val_loss={val_metrics['loss']:.4f} | "
            f"val_macro_F1={val_metrics['macro_F1']:.4f} | "
            f"best_val_macro_F1={best_macro_F1:.4f}"
        )

    if best_state_dict is not None:
        model.load_state_dict(best_state_dict)

    return model


def load_and_split_data():
    df = pd.read_csv(DATA_URL)
    df = df[["RequirementText", "class"]].dropna().reset_index(drop=True)
    df["RequirementText"] = df["RequirementText"].astype(str)
    df["class"] = df["class"].astype(str).str.strip().str.upper()
    df = df[df["RequirementText"].str.strip() != ""].reset_index(drop=True)
    df = df[df["class"].map(df["class"].value_counts()) > 1].reset_index(drop=True)
    df["model_text"] = df["RequirementText"]

    print("Original distribution:", dict(df["class"].value_counts()))

    label_encoder = LabelEncoder()
    df["label"] = label_encoder.fit_transform(df["class"])

    train_df, temp_df = train_test_split(
        df,
        test_size=0.30,
        stratify=df["label"],
        random_state=RANDOM_STATE,
    )

    val_df, test_df = train_test_split(
        temp_df,
        test_size=0.50,
        stratify=temp_df["label"],
        random_state=RANDOM_STATE,
    )

    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)
    test_df = test_df.reset_index(drop=True)

    print("Labels:", list(label_encoder.classes_))
    print("Train distribution:", dict(train_df["class"].value_counts()))
    print("Validation distribution:", dict(val_df["class"].value_counts()))
    print("Test distribution:", dict(test_df["class"].value_counts()))
    print("Model text column: RequirementText/raw")

    return train_df, val_df, test_df, label_encoder


def build_loaders(train_frame, val_frame, test_frame, tokenizer):
    train_dataset = RequirementsDataset(train_frame, tokenizer)
    val_dataset = RequirementsDataset(val_frame, tokenizer)
    test_dataset = RequirementsDataset(test_frame, tokenizer)

    train_loader = DataLoader(
        train_dataset,
        batch_size=TRAIN_BATCH_SIZE,
        shuffle=True,
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=EVAL_BATCH_SIZE,
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=EVAL_BATCH_SIZE,
    )
    return train_loader, val_loader, test_loader


def build_model(label_encoder):
    id2label = {idx: label for idx, label in enumerate(label_encoder.classes_)}
    label2id = {label: idx for idx, label in id2label.items()}
    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=len(label_encoder.classes_),
        id2label=id2label,
        label2id=label2id,
    )
    # Freeze lower transformer layers to reduce overfitting on short requirement texts.
    for layer in model.electra.encoder.layer[:8]:
        for param in layer.parameters():
            param.requires_grad = False

    trainable_params = sum(param.numel() for param in model.parameters() if param.requires_grad)
    frozen_params = sum(param.numel() for param in model.parameters() if not param.requires_grad)
    print(f"Trainable parameters: {trainable_params:,}")
    print(f"Frozen parameters: {frozen_params:,}")
    return model


def evaluate_and_plot(model, test_loader, device, label_encoder, run_name):
    test_metrics, test_labels, test_preds = evaluate_model(model, test_loader, device)

    print("\nTest metrics:")
    print(test_metrics)

    print("\nClassification report:")
    print(classification_report(
        test_labels,
        test_preds,
        target_names=label_encoder.classes_,
        zero_division=0,
    ))

    cm = confusion_matrix(test_labels, test_preds)
    plt.figure(figsize=(10, 7))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_,
        cmap="Blues",
    )
    plt.title(f"ELECTRA Confusion Matrix - {run_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

    return test_metrics, test_labels, test_preds


train_df, val_df, test_df, label_encoder = load_and_split_data()

# --- Augment train split only ---
augmented_rows = []
train_counts = train_df["class"].value_counts().to_dict()

for cls, target_count in AUGMENTATION_TARGETS.items():
    current_count = train_counts.get(cls, 0)
    needed = max(0, target_count - current_count)

    if needed == 0:
        continue

    class_df = train_df[train_df["class"] == cls].reset_index(drop=True)
    if class_df.empty:
        continue

    for _ in range(needed):
        row = class_df.sample(n=1, random_state=random.randint(0, 10**9)).iloc[0]
        aug_text = augment_requirement_text(row["RequirementText"])
        augmented_rows.append({
            "RequirementText": aug_text,
            "class": cls,
            "model_text": aug_text,
            "label": int(row["label"]),
        })

df_aug = pd.DataFrame(augmented_rows)
train_used_df = (
    pd.concat([train_df, df_aug], ignore_index=True)
    if not df_aug.empty
    else train_df.copy()
)
train_used_df = train_used_df.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)

print("Augmented rows added:", len(df_aug))
print("Train after augmentation:", dict(train_used_df["class"].value_counts()))

# --- Tokenizer, datasets, loaders ---
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=True)
train_loader, val_loader, test_loader = build_loaders(train_used_df, val_df, test_df, tokenizer)

# --- Model ---
model = build_model(label_encoder)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

class_weights = compute_class_weights(train_used_df, label_encoder)

model = train_electra(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    device=device,
    class_weights=class_weights,
)

electra_aug_metrics, electra_aug_labels, electra_aug_preds = evaluate_and_plot(
    model,
    test_loader,
    device,
    label_encoder,
    "With Augmentation",
)
```

```python
# --- ELECTRA WITHOUT Augmentation ---
import os
os.environ["WANDB_DISABLED"] = "true"
os.environ["TRANSFORMERS_NO_TF"] = "1"
os.environ["TRANSFORMERS_NO_FLAX"] = "1"

import copy
import random
import re

import nltk
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import seaborn as sns

from nltk.corpus import stopwords, wordnet
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support,
)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from torch.utils.data import DataLoader, Dataset
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
    get_linear_schedule_with_warmup,
)

# --- Config ---
RANDOM_STATE = 42
MODEL_NAME = "google/electra-base-discriminator"
DATA_URL = "https://raw.githubusercontent.com/egbertbritannia-cpu/rbl-requirements-classification/main/data/exp/promise_exp.csv"

MAX_LENGTH = 128
NUM_EPOCHS = 30
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
TRAIN_BATCH_SIZE = 16
EVAL_BATCH_SIZE = 32
WARMUP_RATIO = 0.1

AUGMENTATION_TARGETS = {
    "FT": 200,
    "L": 200,
    "MN": 200,
    "A": 200,
    "LF": 200,
    "SC": 200,
    "PO": 200,
    "PE": 200,
    "O": 200,
    "US": 200,
    "SE": 100,
}

KEEP_WORDS = {"not", "no", "shall", "must", "should", "will", "cannot", "never", "only"}

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
stop_words = set(stopwords.words("english")) - KEEP_WORDS


def get_synonyms(word):
    clean_word = re.sub(r"[^a-zA-Z]", "", str(word).lower())
    if not clean_word:
        return []
    synonyms = set()
    for syn in wordnet.synsets(clean_word):
        for lemma in syn.lemmas():
            synonym = lemma.name().replace("_", " ").lower()
            if synonym != clean_word:
                synonyms.add(synonym)
    return list(synonyms)


def synonym_replacement(text, n=1):
    words = str(text).split()
    if not words:
        return str(text)

    new_words = words.copy()
    candidates = list({
        word for word in words
        if re.sub(r"[^a-zA-Z]", "", word).lower() not in stop_words and len(word) > 3
    })
    random.shuffle(candidates)

    replaced = 0
    for word in candidates:
        synonyms = get_synonyms(word)
        if synonyms:
            synonym = random.choice(synonyms)
            new_words = [synonym if current == word else current for current in new_words]
            replaced += 1
        if replaced >= n:
            break

    return " ".join(new_words)


def phrase_insertion(text, n=1):
    phrases = ["the system shall", "must be able to", "should ensure that"]
    words = str(text).split()
    if not words:
        return str(text)

    for _ in range(n):
        insert_at = random.randint(0, len(words) - 1)
        words.insert(insert_at, random.choice(phrases))

    return " ".join(words)


def random_swap(text, n=1):
    words = str(text).split()
    if len(words) < 2:
        return str(text)

    for _ in range(n):
        i, j = random.sample(range(len(words)), 2)
        words[i], words[j] = words[j], words[i]

    return " ".join(words)


def random_deletion(text, p=0.1):
    words = str(text).split()
    if len(words) <= 1:
        return str(text)

    remaining = [word for word in words if random.random() > p]
    return " ".join(remaining) if remaining else random.choice(words)


def augment_requirement_text(text):
    text = synonym_replacement(text, n=1)
    text = phrase_insertion(text, n=1)
    text = random_swap(text, n=1)
    text = random_deletion(text, p=0.1)
    return " ".join(str(text).split())


class RequirementsDataset(Dataset):
    def __init__(self, frame, tokenizer):
        self.labels = frame["label"].astype(int).tolist()
        self.encodings = tokenizer(
            frame["model_text"].astype(str).tolist(),
            truncation=True,
            padding="max_length",
            max_length=MAX_LENGTH,
        )

    def __getitem__(self, idx):
        item = {key: torch.tensor(values[idx]) for key, values in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

    def __len__(self):
        return len(self.labels)


def compute_metrics_from_arrays(labels, preds):
    weighted_p, weighted_r, weighted_f1, _ = precision_recall_fscore_support(
        labels, preds, average="weighted", zero_division=0
    )
    macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
        labels, preds, average="macro", zero_division=0
    )
    return {
        "accuracy": accuracy_score(labels, preds),
        "precision": weighted_p,
        "recall": weighted_r,
        "f1": weighted_f1,
        "precision_macro": macro_p,
        "recall_macro": macro_r,
        "macro_F1": macro_f1,
    }


@torch.no_grad()
def evaluate_model(model, data_loader, device):
    model.eval()
    all_labels = []
    all_preds = []
    total_loss = 0.0

    for batch in data_loader:
        batch = {key: value.to(device) for key, value in batch.items()}
        outputs = model(**batch)
        total_loss += outputs.loss.item()

        preds = outputs.logits.argmax(dim=-1)
        all_labels.extend(batch["labels"].detach().cpu().numpy())
        all_preds.extend(preds.detach().cpu().numpy())

    metrics = compute_metrics_from_arrays(np.array(all_labels), np.array(all_preds))
    metrics["loss"] = total_loss / max(1, len(data_loader))
    return metrics, np.array(all_labels), np.array(all_preds)


def compute_class_weights(frame, label_encoder):
    classes = np.arange(len(label_encoder.classes_))
    weights = compute_class_weight(
        class_weight="balanced",
        classes=classes,
        y=frame["label"].astype(int).to_numpy(),
    )
    return torch.tensor(weights, dtype=torch.float)


def train_electra(model, train_loader, val_loader, device, class_weights):
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
    )

    total_steps = max(1, NUM_EPOCHS * len(train_loader))
    warmup_steps = int(WARMUP_RATIO * total_steps)

    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_steps,
    )

    loss_fct = torch.nn.CrossEntropyLoss(weight=class_weights.to(device))
    best_macro_F1 = -1.0
    best_state_dict = None

    for epoch in range(1, NUM_EPOCHS + 1):
        model.train()
        total_train_loss = 0.0

        for batch in train_loader:
            batch = {key: value.to(device) for key, value in batch.items()}
            labels = batch.pop("labels")

            optimizer.zero_grad(set_to_none=True)
            outputs = model(**batch)
            loss = loss_fct(outputs.logits, labels)

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            scheduler.step()

            total_train_loss += loss.item()

        val_metrics, _, _ = evaluate_model(model, val_loader, device)
        avg_train_loss = total_train_loss / max(1, len(train_loader))

        if val_metrics["macro_F1"] > best_macro_F1:
            best_macro_F1 = val_metrics["macro_F1"]
            best_state_dict = copy.deepcopy({
                key: value.detach().cpu()
                for key, value in model.state_dict().items()
            })

        print(
            f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
            f"train_loss={avg_train_loss:.4f} | "
            f"val_loss={val_metrics['loss']:.4f} | "
            f"val_macro_F1={val_metrics['macro_F1']:.4f} | "
            f"best_val_macro_F1={best_macro_F1:.4f}"
        )

    if best_state_dict is not None:
        model.load_state_dict(best_state_dict)

    return model


def load_and_split_data():
    df = pd.read_csv(DATA_URL)
    df = df[["RequirementText", "class"]].dropna().reset_index(drop=True)
    df["RequirementText"] = df["RequirementText"].astype(str)
    df["class"] = df["class"].astype(str).str.strip().str.upper()
    df = df[df["RequirementText"].str.strip() != ""].reset_index(drop=True)
    df = df[df["class"].map(df["class"].value_counts()) > 1].reset_index(drop=True)
    df["model_text"] = df["RequirementText"]

    print("Original distribution:", dict(df["class"].value_counts()))

    label_encoder = LabelEncoder()
    df["label"] = label_encoder.fit_transform(df["class"])

    train_df, temp_df = train_test_split(
        df,
        test_size=0.30,
        stratify=df["label"],
        random_state=RANDOM_STATE,
    )

    val_df, test_df = train_test_split(
        temp_df,
        test_size=0.50,
        stratify=temp_df["label"],
        random_state=RANDOM_STATE,
    )

    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)
    test_df = test_df.reset_index(drop=True)

    print("Labels:", list(label_encoder.classes_))
    print("Train distribution:", dict(train_df["class"].value_counts()))
    print("Validation distribution:", dict(val_df["class"].value_counts()))
    print("Test distribution:", dict(test_df["class"].value_counts()))
    print("Model text column: RequirementText/raw")

    return train_df, val_df, test_df, label_encoder


def build_loaders(train_frame, val_frame, test_frame, tokenizer):
    train_dataset = RequirementsDataset(train_frame, tokenizer)
    val_dataset = RequirementsDataset(val_frame, tokenizer)
    test_dataset = RequirementsDataset(test_frame, tokenizer)

    train_loader = DataLoader(
        train_dataset,
        batch_size=TRAIN_BATCH_SIZE,
        shuffle=True,
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=EVAL_BATCH_SIZE,
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=EVAL_BATCH_SIZE,
    )
    return train_loader, val_loader, test_loader


def build_model(label_encoder):
    id2label = {idx: label for idx, label in enumerate(label_encoder.classes_)}
    label2id = {label: idx for idx, label in id2label.items()}
    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=len(label_encoder.classes_),
        id2label=id2label,
        label2id=label2id,
    )
    # Freeze lower transformer layers to reduce overfitting on short requirement texts.
    for layer in model.electra.encoder.layer[:8]:
        for param in layer.parameters():
            param.requires_grad = False

    trainable_params = sum(param.numel() for param in model.parameters() if param.requires_grad)
    frozen_params = sum(param.numel() for param in model.parameters() if not param.requires_grad)
    print(f"Trainable parameters: {trainable_params:,}")
    print(f"Frozen parameters: {frozen_params:,}")
    return model


def evaluate_and_plot(model, test_loader, device, label_encoder, run_name):
    test_metrics, test_labels, test_preds = evaluate_model(model, test_loader, device)

    print("\nTest metrics:")
    print(test_metrics)

    print("\nClassification report:")
    print(classification_report(
        test_labels,
        test_preds,
        target_names=label_encoder.classes_,
        zero_division=0,
    ))

    cm = confusion_matrix(test_labels, test_preds)
    plt.figure(figsize=(10, 7))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_,
        cmap="Blues",
    )
    plt.title(f"ELECTRA Confusion Matrix - {run_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

    return test_metrics, test_labels, test_preds


train_df, val_df, test_df, label_encoder = load_and_split_data()

train_used_df = train_df.copy()
print("Train used without augmentation:", dict(train_used_df["class"].value_counts()))

# --- Tokenizer, datasets, loaders ---
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=True)
train_loader, val_loader, test_loader = build_loaders(train_used_df, val_df, test_df, tokenizer)

# --- Model ---
model = build_model(label_encoder)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

class_weights = compute_class_weights(train_used_df, label_encoder)

model = train_electra(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    device=device,
    class_weights=class_weights,
)

electra_no_aug_metrics, electra_no_aug_labels, electra_no_aug_preds = evaluate_and_plot(
    model,
    test_loader,
    device,
    label_encoder,
    "No Augmentation",
)
```
